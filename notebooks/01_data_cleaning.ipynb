```python
covid.columns = covid.columns.str.lower().str.replace(" ", "_")
worldometer.columns = worldometer.columns.str.lower().str.replace(" ", "_")
```

```python
covid.rename(columns = {"country/region": "country"}, inplace = True)
worldometer.rename(columns = {"country/region": "country"}, inplace = True)
```

```python
covid["date"] = pd.to_datetime(covid["date"])
```

```python
worldometer = worldometer[["country", "continent", "population"]].copy()
```

```python
worldometer = worldometer[worldometer["country"] != "Diamond Princess"].copy()
```

```python
covid["country"] = covid["country"].str.strip()
worldometer["country"] = worldometer["country"].str.strip()
```

```python
print("COVID:", covid["country"].nunique())
print("Worldometer:", worldometer["country"].nunique())
```

```python
country_mapping = {
    "Myanmar": "Burma",
    "CAR" : "Central African Republic",
    "Congo": "Congo (Brazzaville)",
    "DRC" : "Congo (Kinshasa)",
    "Ivory Coast": "Cote d'Ivoire",
    "Vatican City": "Holy See",
    "St. Vincent Grenadines": "Saint Vincent and the Grenadines",
    "S. Korea": "South Korea",
    "Taiwan": "Taiwan*",
    "USA": "US",
    "UAE": "United Arab Emirates",
    "UK": "United Kingdom",
    "Palestine": "West Bank and Gaza"
}
```

```python
worldometer["country"] = worldometer["country"].replace(country_mapping)
```

```python
numeric_columns = covid.select_dtypes(include = "number")
(numeric_columns < 0).sum()
```

```python
covid[covid["active"] < 0]
```

```python
expected_active = covid["confirmed"] - covid["deaths"] - covid["recovered"]
(covid["active"] != expected_active).sum()
```

```python
covid.loc[covid["active"] < 0, "active"] = np.nan
```

```python
covid["active"].isnull().sum()
```

```python
covid_merged = pd.merge(
    covid,
    worldometer,
    on = "country",
    how = "left",
    validate = "many_to_one"
)
```

```python
covid_merged.head()
```

```python
covid_merged.shape
```

```python
covid_merged.isnull().sum()
```

```python
covid_merged.loc[
    covid_merged["population"].isnull(), "country"
].unique()
```

```python
covid_merged.loc[
    covid_merged["country"] == "China", "continent"
] = "Asia" 

covid_merged.loc[
    covid_merged["country"] == "Kosovo", "continent" 
] = "Europe" 
```

```python
covid_merged.isnull().sum() 
```

```python
covid_merged.to_csv(
    "../data/processed/covid_cleaned.csv",
    index = False 
)
```
